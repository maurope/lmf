```r
colnames(df)
```

```r
total_batch_list <- sort(unique(df$batch))

batch_groups <- split(
  total_batch_list,
  ceiling(seq_along(total_batch_list) / 20)
)
```

```r
plots <- lapply(seq_along(batch_groups), function(i) {

  df <- df %>%
    filter(batch %in% batch_groups[[i]])

  ggplot(df, aes(x = factor(batch), y = net_gas_24h_ml)) +
    geom_boxplot() +
    coord_flip() +
    theme_classic() +
    labs(title = paste("Batches", i))
})
```
